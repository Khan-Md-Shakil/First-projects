```python
#H1 Write a function to calculate the sum of numbers from 1 to N.
```

```python
def calculate_sum(n):
    total = 0

    for i in range(1, n + 1):
        total += i

    return total


n = int(input("Enter a number: "))
result = calculate_sum(n)

print("Sum:", result)
```

```python
#H4 Write a function that returns the larger number from two given numbers.
```

```python
def larger_number(a, b):
    if a > b:
        return a
    else:
        return b


num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

result = larger_number(num1, num2)

print("Larger number:", result)
```

```python
#H6 Write a program to find all unique elements in the list.
```

```python
numbers = [10, 20, 10, 30, 40, 20, 50, 30, 60]

unique = []

for number in numbers:
    if number not in unique:
        unique.append(number)

print("Unique elements:", unique)
```

```python
#H7 Write a Python program to calculate the mean of a list of numbers.
```

```python
numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    total += number

mean = total / len(numbers)

print("Mean:", mean)
```

```python
#H8 Classify Numbers into Even and Odd Lists
```

```python
numbers = [10, 15, 20, 25, 30, 35, 40]

even_numbers = []
odd_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)

print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
```

```python
## String Functions

# Initial string
s = "Skill Morph"

# 1. upper()
print(s.upper())  # Converts all characters to uppercase
# 2. lower()
print(s.lower())  # Converts all characters to lowercase
# 3. partition()
print(s.partition(' '))  # Splits the string at the first occurrence of ' '
# 4. replace()
print(s.replace('Morph', 'Master'))  # Replaces 'Morph' with 'Master'
# 5. find()
print(s.find('Morph'))  # Returns the index of 'Morph' in the string, or -1 if not found
# 6. rstrip()
print(s.rstrip())  # Removes trailing spaces from the string
# 7. split()
print(s.split())  # Splits the string into a list at spaces
# 8. startswith()
print(s.startswith('Skill'))  # Checks if the string starts with 'Skill'
# 9. isnumeric()
print(s.isnumeric())  # Checks if all characters in the string are numeric (returns False)
# 10. index()
print(s.index('Morph'))  # Returns the index of the first occurrence of 'Morph'
```
